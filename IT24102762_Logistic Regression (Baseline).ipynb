For our baseline, I implemented the Logistic Regression model.
We used BERT-based sentence embeddings to represent each post as a 384-dimensional vector.
This model is fast and interpretable, making it a good starting point.
It achieved an overall accuracy of around 72% with balanced F1-scores across all classes.
From the confusion matrix, we saw it performed especially well in detecting Bipolar and Anxiety categories.
This baseline helped us understand the dataset’s separability and guided our tuning for more complex models later.

```python
import pandas as pd, numpy as np, joblib
from sentence_transformers import SentenceTransformer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

# 1️Load Data
df = pd.read_csv("preprocessed_data.csv").dropna(subset=["processed_text","target"]).reset_index(drop=True)
X = df["processed_text"].tolist()
y = df["target"]

# 2️ Embeddings
model = SentenceTransformer("all-MiniLM-L6-v2")
X_emb = model.encode(X, batch_size=32, show_progress_bar=True)

# 3️ Split
X_train, X_test, y_train, y_test = train_test_split(X_emb, y, test_size=0.2, random_state=42, stratify=y)

# 4️ Model
clf = LogisticRegression(max_iter=2000, multi_class="multinomial", random_state=42)
clf.fit(X_train, y_train)

# 5️ Evaluate
y_pred = clf.predict(X_test)
print("=== Logistic Regression (Baseline) ===")
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```
